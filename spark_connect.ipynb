# Local Spark Connect

One set of pure PySpark / Spark SQL checks, run from the laptop against a Spark Connect endpoint:
- **`sail`**: a local Sail server started in-process, with tables in OneLake through Sail's Iceberg catalog (same setup as `platforms/lakesail/run.py`). Needs `az login` and `ONELAKE_WAREHOUSE` in `.env`.
- **`snowflake`**: Snowflake's Spark Connect server running on the laptop (`platforms/snowflake/local_server.py`, `sc://localhost:15002`), executing on a Snowflake warehouse. The notebook starts it if it isn't running. Needs the `SNOWFLAKE_*` keys in `.env`.
- **`fabric`**: Fabric Spark through a Livy session. Fabric has no Spark Connect endpoint, so after the Connect cell every cell is sent unchanged to the remote session and its output printed here, except cells whose first line is `# laptop`. Tables are Delta. Needs `az login` and `FABRIC_SPARK_LAKEHOUSE` in `.env` (the same lakehouse as `platforms/fabric/run.py`).
- any `sc://...` URL.

**Kernel: `.venv-snowflake`** (the Spark 3.5.6 client). Snowflake's server doesn't accept Spark 4 clients: every `spark.sql()` arrives empty. Sail accepts both.

    python -m venv .venv-snowflake
    .venv-snowflake/Scripts/python -m pip install --no-cache-dir "snowpark-connect[jdk]" pysail azure-identity ipykernel
    echo snowflake/snowpark_connect/includes/python > .venv-snowflake/Lib/site-packages/vendored_pyspark.pth

Every check prints `PASS`, `WRONG` (ran, bad answer) or `FAIL` (error), so *Run All* goes to the end. A failure is the finding: no workarounds (see `Agent.md`).

```python
# laptop
import os
ENDPOINT = os.environ.get("NB_ENDPOINT", "sail")   # "sail", "snowflake", "fabric", or any "sc://host:port/;token=..." URL
FORMAT = "delta" if ENDPOINT == "fabric" else "iceberg"
SCHEMA = {"sail": "local_nb", "snowflake": "TPCH.LOCAL_NB"}.get(ENDPOINT, "local_nb")
```

## Connect
The only endpoint-specific cell.

```python
# laptop
import os, socket, subprocess, time
from pathlib import Path

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / ".env").exists())
for line in open(ROOT / ".env"):
    if "=" in line and not line.lstrip().startswith("#"):
        k, v = line.split("=", 1)
        os.environ.setdefault(k.strip(), v.strip().strip('"').strip("'"))

def port_open(port):
    with socket.socket() as s:
        return s.connect_ex(("127.0.0.1", port)) == 0

# Cells go to a Livy session only while ENDPOINT is fabric: drop the hook a previous fabric run installed.
ip = get_ipython()
ip.input_transformers_cleanup[:] = [t for t in ip.input_transformers_cleanup if getattr(t, "__name__", "") != "to_livy"]

server = proc = None
if ENDPOINT == "sail":
    # One Entra token for https://storage.azure.com/ covers both the catalog and the files: bearer_token for the
    # catalog, AZURE_STORAGE_TOKEN for storage (Sail has no credential vending). Read once, at server start.
    from azure.identity import AzureCliCredential
    token = AzureCliCredential().get_token("https://storage.azure.com/.default").token
    os.environ["SAIL_CATALOG__LIST"] = (
        f'[{{type="onelake", name="onelake", url="{os.environ["ONELAKE_WAREHOUSE"]}", '
        f'api="iceberg", bearer_token="{token}", '
        f'table_cache_type="session", table_cache_ttl_secs=300, '
        f'database_cache_type="session", database_cache_ttl_secs=300}}]'
    )
    os.environ["SAIL_CATALOG__DEFAULT_CATALOG"] = "onelake"
    os.environ["AZURE_STORAGE_TOKEN"] = token
    os.environ.setdefault("RUST_LOG", "error")
    from pysail.spark import SparkConnectServer
    server = SparkConnectServer()
    server.start()
    url = f"sc://localhost:{server.listening_address[1]}"
elif ENDPOINT == "snowflake":
    if not port_open(15002):
        log = open(ROOT / ".venv-snowflake" / "server.log", "w")
        proc = subprocess.Popen([str(ROOT / ".venv-snowflake" / "Scripts" / "python.exe"),
                                 str(ROOT / "platforms" / "snowflake" / "local_server.py")],
                                stdout=log, stderr=subprocess.STDOUT)
        while not port_open(15002):
            if proc.poll() is not None:
                raise RuntimeError("Snowflake server exited, see .venv-snowflake/server.log")
            time.sleep(1)
    url = "sc://localhost:15002"
elif ENDPOINT == "fabric":
    # Fabric has no Spark Connect endpoint. A Livy session is its interactive path: a warm remote Spark session
    # that runs code strings. Modelled on dbt-fabricspark's Livy client (same CLI token scope).
    import json, requests
    from azure.identity import AzureCliCredential
    ws, lh = os.environ["FABRIC_SPARK_LAKEHOUSE"].split("/")
    url = f"https://api.fabric.microsoft.com/v1/workspaces/{ws}/lakehouses/{lh}/livyapi/versions/2023-12-01/sessions"
    cred, tok = AzureCliCredential(), None

    def http(method, path="", **kw):
        global tok
        if tok is None or tok.expires_on - time.time() < 300:
            tok = cred.get_token("https://analysis.windows.net/powerbi/api/.default")
        for attempt in range(6):
            try:
                r = requests.request(method, url + path, headers={"Authorization": f"Bearer {tok.token}"},
                                     timeout=60, **kw)
                if r.status_code != 429 and r.status_code < 500:
                    break
            except (requests.ConnectionError, requests.Timeout):
                if attempt == 5:
                    raise
            time.sleep(min(2 ** attempt, 30))
        if not r.ok:
            raise RuntimeError(f"Livy {method} {path or '/'} -> {r.status_code} {r.text[:500]}")
        return r.json() if r.content else {}

    def livy(code, show=True):
        """Run code in the remote session, print its output. A remote exception is raised here."""
        st, wait = http("POST", f"/{sid}/statements", json={"code": code, "kind": "pyspark"}), 0.2
        while st.get("state") not in ("available", "error", "cancelled"):
            time.sleep(wait)
            wait = min(wait * 1.5, 2)
            st = http("GET", f"/{sid}/statements/{st['id']}")
        out = st.get("output") or {}
        if out.get("status") != "ok":
            print("".join(out.get("traceback") or []), end="")
            raise RuntimeError(f"remote {out.get('ename', st.get('state'))}: {out.get('evalue', '')}")
        text = out.get("data", {}).get("text/plain", "")
        if show and text:
            print(text)
        return text

    t0 = time.time()
    sid = http("POST", json={"conf": {"spark.sql.session.timeZone": "UTC",
                                      "spark.sql.ansi.enabled": "false"}})["id"]   # Spark 3.5 semantics, as the jobs expect
    while True:
        s = http("GET", f"/{sid}")
        top, cur = s.get("state"), (s.get("livyInfo") or {}).get("currentState")
        if top == "idle" or (cur == "idle" and top not in ("starting", "not_started")):
            break
        if {top, cur} & {"dead", "error", "killed", "shutting_down"}:
            raise RuntimeError(f"Livy session {sid}: state {top}/{cur}")
        time.sleep(3)
    print(f"Livy session {sid} idle after {time.time() - t0:.0f}s")

    # The remote namespace gets what the cells use.
    livy(f"import time\nENDPOINT, FORMAT, SCHEMA = {ENDPOINT!r}, {FORMAT!r}, {SCHEMA!r}")

    def to_livy(lines):
        """From the next cell on: any cell not starting with '# laptop' runs in the Livy session."""
        return lines if "".join(lines).lstrip().startswith("# laptop") else [f"livy({''.join(lines)!r})\n"]
    ip.input_transformers_cleanup.append(to_livy)

    CLIENT = "livy"
    ENGINE_VERSION = livy("print(spark.sql('SELECT version()').first()[0])", show=False).strip()
    pull = lambda name: json.loads(livy(f"import json; print(json.dumps({name}))", show=False))
    close = lambda: http("DELETE", f"/{sid}")
    print(f"fabric: Livy session {sid} | engine {ENGINE_VERSION} | "
          f"spark.version {livy('print(spark.version)', show=False).strip()}")
else:
    url = ENDPOINT

if ENDPOINT != "fabric":
    import pyspark
    from pyspark.sql import SparkSession
    spark = SparkSession.builder.remote(url).getOrCreate()
    spark.conf.set("spark.sql.session.timeZone", "UTC")
    if ENDPOINT == "sail":
        spark.conf.set("spark.sql.ansi.enabled", "false")      # Spark 3.5 semantics, as the jobs expect
    if ENDPOINT == "snowflake":
        spark.conf.set("snowpark.connect.iceberg.external_volume", "SNOWFLAKE_MANAGED")
    CLIENT, ENGINE_VERSION = f"pyspark {pyspark.__version__}", spark.sql("SELECT version()").first()[0]
    pull = lambda name: globals()[name]

    def close():
        spark.stop()
        if server:
            server.stop()
        if proc:                  # only a Snowflake server this notebook started
            proc.terminate()
    print(f"{ENDPOINT}: {url.split(';')[0]} | client {spark.version} | engine {ENGINE_VERSION}")
```

## Checks

```python
import datetime as dt
import pandas as pd
from pyspark.sql import Window, functions as F

RESULTS = []
_MISSING = object()

def check(name, expect=_MISSING):
    """Run the decorated function now and record PASS / WRONG / FAIL. With no expect, running without error is a PASS."""
    def run(fn):
        t0, got = time.time(), None
        try:
            got = fn()
            ok = expect is _MISSING or got == expect
            status, detail = ("PASS", "") if ok else ("WRONG", f"got {got!r}, expected {expect!r}")
        except Exception as e:
            status, detail = "FAIL", f"{type(e).__name__}: {' '.join(str(e).split())[:300]}"
        RESULTS.append(dict(check=name, status=status, secs=round(time.time() - t0, 1), detail=detail))
        print(f"{status:5} {name} ({time.time() - t0:.1f}s) {detail}")
        return got
    return run

def rows(df):
    return [tuple(r) for r in df.collect()]

ORDERS = [(1, "a", "AU", 10.0, "2026-09-01"), (2, "b", "AU", 20.0, "2026-09-02"),
          (3, "a", "NZ", 5.0, "2026-09-02"), (4, "c", "US", None, "2026-09-03")]
SCHEMA_DDL = "id int, cust string, country string, amount double, day string"
T, T_PART = f"{SCHEMA}.nb_orders", f"{SCHEMA}.nb_orders_part"
```

### Catalog

```python
@check("CREATE SCHEMA IF NOT EXISTS")
def _(): spark.sql(f"CREATE SCHEMA IF NOT EXISTS {SCHEMA}")

@check("USE SCHEMA")
def _(): spark.sql(f"USE SCHEMA {SCHEMA}")

@check("SHOW NAMESPACES lists it", True)
def _(): return SCHEMA.split(".")[-1].lower() in [r[0].lower() for r in spark.sql("SHOW NAMESPACES").collect()]

@check("catalog.currentDatabase", SCHEMA.split(".")[-1].lower())
def _(): return spark.catalog.currentDatabase().lower()

@check("catalog.listTables")
def _(): spark.catalog.listTables()
```

### DataFrame API

```python
orders = spark.createDataFrame(ORDERS, SCHEMA_DDL)
custs = spark.createDataFrame([("a", "Alice"), ("b", "Bob")], "cust string, name string")

@check("createDataFrame (rows)", 4)
def _(): return orders.count()

@check("createDataFrame (pandas)", 6)
def _(): return spark.createDataFrame(pd.DataFrame({"x": [1, 2, 3]})).agg(F.sum("x")).first()[0]

@check("range + filter", 5)
def _(): return spark.range(10).filter("id % 2 = 0").count()

@check("withColumn + select", 40.0)
def _(): return orders.withColumn("twice", F.col("amount") * 2).where("id = 2").select("twice").first()[0]

@check("groupBy + agg", [("AU", 2, 30.0), ("NZ", 1, 5.0), ("US", 1, None)])
def _(): return rows(orders.groupBy("country").agg(F.count("*"), F.sum("amount")).orderBy("country"))

@check("join inner / left / anti", (3, 4, 1))
def _(): return tuple(orders.join(custs, "cust", how).count() for how in ("inner", "left", "left_anti"))

@check("union + distinct", 3)
def _(): return orders.select("country").union(orders.select("country")).distinct().count()

@check("dropDuplicates", 3)
def _(): return orders.dropDuplicates(["cust"]).count()

@check("orderBy desc + limit", [4, 3])
def _(): return [r[0] for r in orders.orderBy(F.desc("id")).limit(2).collect()]

@check("pivot", [("a", 10.0, 5.0), ("b", 20.0, None), ("c", None, None)])
def _(): return rows(orders.groupBy("cust").pivot("country", ["AU", "NZ"]).agg(F.sum("amount")).orderBy("cust"))

@check("fillna", 35.0)
def _(): return orders.fillna({"amount": 0.0}).agg(F.sum("amount")).first()[0]

@check("explode", 3)
def _(): return spark.createDataFrame([(1, [1, 2, 3])], "id int, xs array<int>").select(F.explode("xs")).count()

@check("window row_number", 3)
def _(): return orders.withColumn("rn", F.row_number().over(Window.partitionBy("country").orderBy("id"))).where("rn = 1").count()

@check("toPandas", 4)
def _(): return len(orders.toPandas())
```

### Spark SQL

```python
orders.createOrReplaceTempView("v_orders")

@check("temp view", 4)
def _(): return spark.sql("SELECT COUNT(*) FROM v_orders").first()[0]

@check("CTE", 2)
def _(): return spark.sql("WITH t AS (SELECT country, SUM(amount) s FROM v_orders GROUP BY country) "
                          "SELECT COUNT(*) FROM t WHERE s > 0").first()[0]

@check("window running sum", [(1, 10.0), (2, 30.0), (3, 35.0), (4, 35.0)])
def _(): return rows(spark.sql("SELECT id, SUM(amount) OVER (ORDER BY id) FROM v_orders ORDER BY id"))

@check("CASE WHEN", 2)
def _(): return spark.sql("SELECT SUM(CASE WHEN amount >= 10 THEN 1 ELSE 0 END) FROM v_orders").first()[0]

@check("date functions", (dt.date(2026, 10, 1), 115, "2026-09"))
def _(): return tuple(spark.sql("SELECT date_add(to_date('2026-09-01'), 30), "
                                "datediff(to_date('2026-12-25'), to_date('2026-09-01')), "
                                "date_format(to_timestamp('2026-09-01 13:45:00'), 'yyyy-MM')").first())

@check("try_cast", (None, 42))
def _(): return tuple(spark.sql("SELECT try_cast('x' AS INT), try_cast('42' AS INT)").first())

@check("string functions", ("AB", "005", "a-b", "007"))
def _(): return tuple(spark.sql(r"SELECT upper(trim('  ab ')), regexp_extract('SKU005', '(\\d+)', 1), "
                                "concat_ws('-', 'a', 'b'), lpad('7', 3, '0')").first())

@check("arrays and maps", (3, [10, 20, 30], 2, True))
def _(): return tuple(spark.sql("SELECT size(array(1, 2, 3)), transform(array(1, 2, 3), x -> x * 10), "
                                "element_at(map('a', 1, 'b', 2), 'b'), array_contains(array(1, 2), 2)").first())

@check("GROUPING SETS", 4)
def _(): return spark.sql("SELECT country, COUNT(*) FROM v_orders GROUP BY GROUPING SETS ((country), ())").count()
```

### UDFs

```python
@check("Python UDF", 6)
def _():
    plus_one = F.udf(lambda x: x + 1, "bigint")
    return spark.range(3).select(plus_one("id").alias("x")).agg(F.sum("x")).first()[0]

@check("pandas UDF", 70.0)
def _():
    @F.pandas_udf("double")
    def twice(s: pd.Series) -> pd.Series:
        return s * 2
    return orders.select(twice("amount").alias("x")).agg(F.sum("x")).first()[0]

@check("udf.register + SQL", 16)
def _():
    spark.udf.register("sq", lambda x: x * x, "bigint")
    return spark.sql("SELECT sq(4)").first()[0]
```

### Table writes (open format only)
Standard Spark on `FORMAT` tables under `SCHEMA`.

```python
spark.createDataFrame([(1, "a", "AU", 99.0, "2026-09-01"), (6, "e", "US", 1.0, "2026-09-05")], SCHEMA_DDL) \
     .createOrReplaceTempView("v_src")

@check("DROP TABLE IF EXISTS")
def _(): spark.sql(f"DROP TABLE IF EXISTS {T}")

@check("saveAsTable (append, new table)", 4)
def _():
    orders.write.format(FORMAT).mode("append").saveAsTable(T)
    return spark.table(T).count()

@check("DESCRIBE TABLE", ["id", "cust", "country", "amount", "day"])
def _(): return [r[0].lower() for r in spark.sql(f"DESCRIBE TABLE {T}").collect()][:5]

@check("INSERT INTO VALUES", 5)
def _():
    spark.sql(f"INSERT INTO {T} VALUES (5, 'd', 'NZ', 7.0, '2026-09-04')")
    return spark.table(T).count()

@check("INSERT OVERWRITE", 4)
def _():
    spark.sql(f"INSERT OVERWRITE {T} SELECT * FROM v_orders")
    return spark.table(T).count()

@check("MERGE upsert", (5, 99.0))
def _():
    spark.sql(f"MERGE INTO {T} t USING v_src s ON t.id = s.id "
              "WHEN MATCHED THEN UPDATE SET * WHEN NOT MATCHED THEN INSERT *")
    return spark.table(T).count(), spark.table(T).where("id = 1").select("amount").first()[0]

@check("UPDATE", 0.0)
def _():
    spark.sql(f"UPDATE {T} SET amount = 0 WHERE id = 6")
    return spark.table(T).where("id = 6").select("amount").first()[0]

@check("DELETE", 4)
def _():
    spark.sql(f"DELETE FROM {T} WHERE id = 6")
    return spark.table(T).count()

@check("writeTo().append()", 5)
def _():
    orders.where("id = 2").writeTo(T).append()
    return spark.table(T).count()

@check("CREATE TABLE PARTITIONED BY + INSERT", 4)
def _():
    spark.sql(f"DROP TABLE IF EXISTS {T_PART}")
    spark.sql(f"CREATE TABLE {T_PART} (id INT, country STRING, amount DOUBLE) USING {FORMAT} PARTITIONED BY (country)")
    spark.sql(f"INSERT INTO {T_PART} SELECT id, country, amount FROM v_orders")
    return spark.table(T_PART).count()
```

## Summary

```python
# laptop
import datetime as dt
import pandas as pd
summary = pd.DataFrame(pull("RESULTS"))
counts = ", ".join(f"{n} {s}" for s, n in summary.status.value_counts().items()) + f" of {len(summary)}"
print(f"{ENDPOINT}: {counts}")

# The run's record: one row per check, committed under findings/local/.
out = ROOT / "findings" / "local" / f"{ENDPOINT if not ENDPOINT.startswith('sc://') else 'custom'}.csv"
out.parent.mkdir(exist_ok=True)
with open(out, "w", newline="", encoding="utf-8") as f:
    f.write(f"# {dt.datetime.now(dt.timezone.utc):%Y-%m-%dT%H:%M:%SZ} endpoint={ENDPOINT.split(';')[0]} format={FORMAT} "
            f"schema={SCHEMA} client={CLIENT} engine={ENGINE_VERSION} | {counts}\n")
    summary.to_csv(f, index=False)
print("wrote", out.relative_to(ROOT).as_posix())
summary[summary.status != "PASS"]
```
